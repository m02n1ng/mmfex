```python
#get data
x_axis = np.array([1.52446,
1.53767,
1.53904,
1.57117,
1.57172,
1.60222,
1.60509,
1.63076,
1.63231,
1.64608,
1.6487 ,
1.64893,
1.65504,
1.65507,
1.6577 ,
1.66252,
1.66318,
1.66377,
1.67639,
1.68129,
1.69189,
1.70046,
1.70984,
1.71355,
1.72281,
1.72563,
1.73536]
)

f = np.array([0.01157,
2.47280,
2.45198,
0.07769,
0.12019,
0.00079,
0.00091,
0.00927,
0.00835,
0.04489,
1.28248,
1.26898,
0.07459,
0.05188,
0.00013,
0.09397,
0.07181,
0.01597,
0.00892,
0.00269,
0.00163,
0.00186,
0.00661,
0.00571,
0.04835,
0.03314,
0.16936])
```

```python
np.min(x_axis)
np.max(x_axis)
diff = np.max(x_axis) - np.min(x_axis)
```

```python
# set parameters and calculate the spectrum
A = 16193.0311
sigma = 0.0272
E = np.linspace(np.min(x_axis)-0.5*diff,np.max(x_axis)+0.5*diff, 1000)
spectrum = 0
for i in range(0,len(x_axis)):
    spectrum += (f[i]/sigma) * np.exp(-((E-x_axis[i])**2)/(sigma**2))
```

```python
max_spectrum = np.max(spectrum)  # Find the maximum y value
max_E = E[spectrum.argmax()]  # Find the x value corresponding to the maximum y value
#print(max_E, max_spectrum)
hc = 6.626e-34 * 3e8
dist =  hc/(E[spectrum.argmax()] * 1.602e-19) * 1e9 - hc/(1.65 * 1.602e-19) * 1e9
```

```python
plt.plot(E, spectrum/np.max(spectrum), "k")
#plt.vlines(1.65, 0, 1)
plt.vlines(E[spectrum.argmax()], 0, 1, color = "k", linestyle = "--", label = round(hc/(E[spectrum.argmax()] * 1.602e-19) * 1e9, 2))
plt.vlines(1.65, 0, 0.58, color = "grey", linestyle = "--", label = round(hc/(1.65 * 1.602e-19) * 1e9, 2))
plt.hlines(0.4, E[spectrum.argmax()], 1.65, colors='r', label = 'distance (nm) ='+ str(round(dist,1)))
#plt.vlines(x_axis, 0, 3E6, colors='r')
plt.title('LH2 Spectrum (Qy only)')
plt.xlabel('Energy (eV)')
plt.ylabel('Absorption (Normalised)')
plt.legend()
```
